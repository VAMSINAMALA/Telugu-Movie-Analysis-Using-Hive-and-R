```python
import pandas as pd
df = pd.read_csv("TeluguMovies_dataset.csv")
```

```python
df.head()
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values)
```

```python
# Fill missing values in 'Year' with the median year
df.loc[:, 'Year'] = df['Year'].fillna(df['Year'].median())

# Fill missing values in 'Certificate' and 'Genre' with 'Unknown'
df.loc[:, 'Certificate'] = df['Certificate'].fillna('Unknown')
df.loc[:, 'Genre'] = df['Genre'].fillna('Unknown')

# Fill missing values in 'Overview' with 'No overview available'
df.loc[:, 'Overview'] = df['Overview'].fillna('No overview available')
```

```python
# Convert data types for consistency
df['Year'] = df['Year'].astype(int)
df['Runtime'] = df['Runtime'].astype(int)
df['Rating'] = df['Rating'].astype(float)
df['Movie'] = df['Movie'].astype(str)
df['Certificate'] = df['Certificate'].astype(str)
df['Genre'] = df['Genre'].astype(str)
df['Overview'] = df['Overview'].astype(str)
```

```python
# Strip whitespace and standardize the case for string columns
df['Movie'] = df['Movie'].str.strip().str.title()
df['Genre'] = df['Genre'].str.strip().str.title()
df['Certificate'] = df['Certificate'].str.strip().str.upper()
df['Overview'] = df['Overview'].str.strip()
```

```python
# Save the preprocessed data to a CSV file
df.to_csv("Preprocessed_TeluguMovies_Dataset.csv", index=False)
```
